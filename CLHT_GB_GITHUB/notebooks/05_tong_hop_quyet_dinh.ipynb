# Tổng hợp và quyết định

Đọc các bảng đã chạy, không tự tạo lại số liệu. Mỗi quyết định phải nêu điều kiện, kết quả, giới hạn và bước tiếp theo.

```python
from pathlib import Path
import sys
ROOT = Path.cwd()
if not (ROOT / 'clht_code').exists() and (ROOT.parent / 'clht_code').exists():
    ROOT = ROOT.parent
sys.path.insert(0,str(ROOT))
DATA=ROOT/'data'
OUT=ROOT/'ket_qua'
OUT.mkdir(exist_ok=True)
import pandas as pd
base=pd.read_csv(OUT/'ket_qua_mo_hinh_co_so.csv'); opt=pd.read_csv(OUT/'ket_qua_hco_vnls.csv'); display(base.groupby('model').agg(mean_ap=('average_precision_risk','mean'),std_ap=('average_precision_risk','std'),mean_recall=('recall_risk','mean'),mean_f1=('weighted_f1','mean')).sort_values('mean_ap',ascending=False)); display(opt.groupby('method').agg(mean_ap=('average_precision_cv','mean'),std_ap=('average_precision_cv','std'),mean_features=('selected_count','mean'),mean_calls=('objective_calls','mean'),mean_seconds=('wall_seconds','mean')).sort_values('mean_ap',ascending=False))
```

## Mẫu ghi quyết định

Model có AP cao nhưng recall thấp: thử threshold/capacity trước khi đổi model. Optimizer chọn ít cột nhưng chậm: so cùng objective calls và cân nhắc lợi ích. Không dùng test để tune. Không diễn giải feature được chọn nhiều là nguyên nhân.
